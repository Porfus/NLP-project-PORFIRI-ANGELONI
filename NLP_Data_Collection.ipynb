```python
import pandas as pd
from pathlib import Path
```

Process for generate the dataset for each Italian dialect by using the wiki articles in json format

```python
dataset = pd.DataFrame()
```

```python
dialect = 'it' #specify the dialect
path = f"{dialect}_texts/AA/" #specify the folder of the dialect

folder_path = Path(path)

for entry in folder_path.iterdir():
    if entry.is_file():
        df = pd.read_json(path + entry.name, lines=True)
        df['dial'] = dialect
        df = df[['text','dial']]
        dataset = pd.concat([dataset, df[['text', 'dial']]], ignore_index=True)
```

```python
#data cleaning
dataset = dataset[dataset['text'].notna() & (dataset['text'] != '')]
print(f"Datatset rows: {len(dataset)}")
dataset.head()
```

```python
dataset.to_csv('dataset.csv', index=False)
```

NB: To execute the script this file must be in the same directory of the dialect folders
